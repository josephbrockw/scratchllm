```python
import onnxruntime as rt

onnx_model_path = "linear_regression.onnx"
session = rt.InferenceSession(onnx_model_path)
```

```python
import numpy as np

X = np.array([[2.0, 4.0, 6.0, 8.0]], dtype=np.float32)
A = np.array([[0.5], [1.0], [1.5], [2.0]], dtype=np.float32)
B = np.array([[0.1]], dtype=np.float32)
```

```python
result = session.run(
    [session.get_outputs()[0].name],
    {"X": X, "A": A, "B": B}
)
print(f"Output: {result[0]}")
```
